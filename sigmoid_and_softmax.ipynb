```python
import torch
import numpy as np
import matplotlib.pyplot as plt
```

```python
def sigmoid(x):
  return 1/(1+np.exp(-x))
```

```python
x_axes = np.linspace(-10, 10, 100)
```

```python
plt.plot(x_axes, sigmoid(x_axes))
plt.xlabel("logit")
plt.ylabel("sigmoid")
plt.axhline(y=0, color="r", linestyle="--")
plt.axhline(y=1, color="r", linestyle="--")
plt.
plt.grid()
plt.plot()
```

```python
def softmax(z):
  return np.exp(z)/np.exp(z).sum()
```

```python
logits = [2, 5, 1]
```

```python
softmax(logits)
```
